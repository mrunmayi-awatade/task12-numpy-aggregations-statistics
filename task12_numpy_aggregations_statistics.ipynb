```python
# Basic aggregation functions
import numpy as np
numbers = np.array([10, 20, 30, 40, 50])
print("Sum:", np.sum(numbers))
print("Mean:", np.mean(numbers))
print("Median:", np.median(numbers))
print("Minimum:", np.min(numbers))
print("Maximum:", np.max(numbers))
```

```python
# Calculating standard deviation and variance
import numpy as np
numbers = np.array([10, 20, 30, 40, 50])
print("Standard Deviation:", np.std(numbers))
print("Variance:", np.var(numbers))
```

```python
# Aggregation along rows and columns
import numpy as np
data = np.array([[10, 20, 30], [40, 50, 60]])
print("Sum of All Elements:", np.sum(data))
print("Sum of Each Column:", np.sum(data, axis=0))
print("Sum of Each Row:", np.sum(data, axis=1))
```

```python
# Statistical analysis of student marks
import numpy as np
marks = np.array([75, 82, 68, 90, 85, 70, 80])
print("Marks:", marks)
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
print("Standard Deviation:", np.std(marks))
```

```python
# Aggregation on a 2D array
import numpy as np
data = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print("Mean:", np.mean(data))
print("Column Means:", np.mean(data, axis=0))
print("Row Means:", np.mean(data, axis=1))
```

```python
# Calculating percentiles
import numpy as np
marks = np.array([40, 50, 60, 70, 80, 90, 100])
print("25th Percentile:", np.percentile(marks, 25))
print("50th Percentile:", np.percentile(marks, 50))
print("75th Percentile:", np.percentile(marks, 75))
```

```python
# Calculating cumulative sum and product
import numpy as np
numbers = np.array([1, 2, 3, 4, 5])
print("Cumulative Sum:", np.cumsum(numbers))
print("Cumulative Product:", np.cumprod(numbers))
```

```python
# Combining aggregation and statistical functions
import numpy as np
marks = np.array([70, 80, 90, 60, 85])
print("Total:", np.sum(marks))
print("Average:", np.mean(marks))
print("Median:", np.median(marks))
print("Variance:", np.var(marks))
print("Standard Deviation:", np.std(marks))
print("Cumulative Sum:", np.cumsum(marks))
```
